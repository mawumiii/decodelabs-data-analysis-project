```python
#extracting year and month from date column
df['Year'] =df['Date'].dt.year
df['Month'] =df['Date'].dt.month
df['Month_Name'] =df['Date'].dt.month_name()
```

```python
#calculate yearly sales
yearly_sales=df.groupby('Year')['TotalPrice'].sum()
print(yearly_sales)
```

```python
#plot yearly sales
yearly_sales.plot(kind='bar')
plt.title('yearly_sales:2023-2025')
plt.xlabel('Year')
plt.ylabel('TotalPrice')
plt.show()
```

```python
#to identify monthly sales trend
monthly_sales=(
    df.set_index('Date')['TotalPrice']
    .resample('ME')
    .sum()
)
```

```python
monthly_sales.plot(figsize=(12,5))
plt.title('Monthly Sales Trend:2023-2025')
plt.xlabel('Date')
plt.ylabel('TotalPrice')
plt.show()
```

```python
#checking for total sales by product

product_sales=df.groupby('Product')['TotalPrice'].sum()
product_sales.plot(kind='bar')
plt.xlabel('Product')
plt.ylabel('TotalPrice')
plt.show()
```

```python
#checking for outliers and unusual patterns
sns.boxplot(data=df,x='TotalPrice')
plt.title('Sales Outliers Detection')
plt.show()
```
